## How it works

Autark is built from four modules: `autk-db` (spatial data, DuckDB), `autk-compute` (WebGPU computation), `autk-map`
(2D/3D maps) and `autk-plot` (linked charts). They exchange **features**: GeoJSON features with properties.

pyautark gives you Python objects that describe a visualization:

1. **Data** sources: `GeoJSON`, `CSV`, `JSON`, `GeoTIFF`, `OSM`.
2. **Transforms** that derive tables: `SpatialJoin`, `Heatmap`, `Compute.gpgpu`, `Compute.render`.
3. **Views**: `Map` with `Layer`s, and the plots `Scatterplot`, `Histogram` and `Table`.
4. **Links** that connect a plot selection to a map layer.

An `ak.Spec` collects them. `spec.to_dict()` is an **UrbanSpec**, the JSON specification of
[autk-grammar](https://github.com/urban-toolkit/autk-grammar). `spec.widget()` runs that JSON in the notebook with
the Autark modules, in your browser. Nothing is sent to a server, and the same JSON runs in any autk-grammar host.

```
Python objects  ->  UrbanSpec (JSON)  ->  autk-db -> autk-compute -> autk-map + autk-plot
```

## A first map

The smallest spec is one data source and one map with one layer. `ak.GeoJSON` takes a `name` (how the rest of the spec
refers to the table) and either a `url` or inline `values`. A layer refers to a table by name, or by the source object
itself.

````python
neighborhoods = ak.GeoJSON("neighborhoods", url=t.url("data/map/mnt_neighs.geojson"))

spec = ak.Spec(
    data=[neighborhoods],
    views=[ak.Map(layers=[ak.Layer(neighborhoods).style(color="#4c78a8", opacity=0.8)])],
)
spec.widget()
````

Drag to pan, scroll to zoom, Shift + drag to rotate and tilt. The button in the top-left corner opens the layer menu.

## The spec is plain JSON

`to_dict()` shows what the widget actually runs: an UrbanSpec. Python names are renamed on the way
(`name` becomes `outputTableName`, `url` becomes `geojsonFileUrl`, a layer's source becomes `dataRef`).

````python
import json
print(json.dumps(spec.to_dict(), indent=2))
````

You can save it, validate it against the grammar's schema, or export a standalone page:

```python
spec.save_json("map.json")
spec.validate()             # needs `pip install jsonschema`
spec.save_html("map.html")  # one file that opens anywhere
```

The data of an exported page must be reachable from the page: give absolute URLs, or inline `values`.
